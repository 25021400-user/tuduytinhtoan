```python
#w7a22
n = int(input("Nhập số lượng sinh viên: "))
danh_sach = []
dem = {}

for _ in range(n):
    ho_ten = input("Nhập họ và tên: ")
    ten = ho_ten.split()[-1]
    danh_sach.append(ho_ten)

    if ten in dem:
        dem[ten] += 1
    else:
        dem[ten] = 1

ten_hay_gap = None
max_count = 0

for ten, count in dem.items():
    if count > max_count:
        max_count = count
        ten_hay_gap = ten
    elif count == max_count:
        if ten < ten_hay_gap:
            ten_hay_gap = ten
print(f"Tên hay gặp nhất là '{ten_hay_gap}' với {max_count} lần xuất hiện.")
```

```python
#w7a25
a=input()
k=int(input())
c=[]
for i in a:
    if i.isdigit():
        c.append(int(i))

stack=[]
solanxoa=len(c)-k
for i in c:
    if len(stack)==0:
        stack.append(i)
        continue
    if i > stack[-1]:
        solan=0
        for j in reversed(stack):
            if j<i and solan<solanxoa:
                solan+=1
            else:
                break
    
        for j in range(solan):
            stack.pop()
        stack.append(i)
        solanxoa-=solan  
        continue     
    if i<=stack[-1]:
        stack.append(i)
        continue
    if solanxoa==0:
        break
if solanxoa>0:
    stack=stack[:-solanxoa]

b=''
for i in stack:
    b+=str(i)
print(int(b))
      


    
```

```python
#w7a28
def songuyento(a):
    if a < 2: return False
    for i in range(2, int(a**0.5) + 1):
        if a % i == 0:
            return False  
    return True  
n = int(input("Nhập N: "))
a=[]
for i in range(2,n):
    a.append(i)
daylonnhat = []
daythu = []

for i in a:
    
    if i>1 and not songuyento(i):
        daythu.append(i)
    else:
        if len(daythu) > len(daylonnhat):
            daylonnhat = daythu.copy()
        daythu = [] 

if len(daythu) > len(daylonnhat):
    daylonnhat = daythu.copy()
print(a)
print("Dãy hợp số dài nhất là:", daylonnhat)
```

```python
#w7a30
a = list(map(int, input("Nhập dãy số: ").split()))

if not a:
    print("Dãy rỗng")
else:
    n = len(a)
    L = [1] * n
    for i in range(1, n):
        for j in range(0, i):
            if a[j] < a[i]:
                L[i] = max(L[i], L[j] + 1)
    max_len = max(L)
    print("Độ dài của dãy con tăng dần dài nhất là:", max_len)
    lis = []
    current_len = max_len
    for i in range(n - 1, -1, -1):
        if L[i] == current_len:
            lis.append(a[i])
            current_len -= 1
    lis.reverse()
    print("Dãy con tăng dần dài nhất đó là:", lis)
```
