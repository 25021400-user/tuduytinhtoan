```python
#w10a1
try:
    a = int(input("Nhập số nguyên a: "))
    b = int(input("Nhập số nguyên b: "))
    result = a / b
    print(f"{result:.2f}")
except ZeroDivisionError:
    print("Lỗi: Không thể chia cho 0.")
except ValueError:
    print("Lỗi: Vui lòng nhập đúng số nguyên.")
```

```python
#W10A2
try:
    n = int(input("Nhập số lượng phần tử n: "))
    if n <= 0:
        print("n phải là số nguyên dương.")
    else:
        my_list = [int(input(f"Nhập phần tử thứ {i+1}: ")) for i in range(n)]
        print(my_list[n])
except IndexError:
    print("Lỗi: Truy cập chỉ số sai (IndexError).")
except ValueError:
    print("Lỗi: Dữ liệu nhập vào không hợp lệ.")
```

```python
#W10A3
def is_happy_number(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(char) ** 2 for char in str(n))
    return n == 1

try:
    a = int(input("Nhập vào một số nguyên dương a: "))
    if a <= 0:
        print("Vui lòng nhập số nguyên dương.")
    else:
        if is_happy_number(a):
            print(f"{a} là số hạnh phúc.")
        else:
            print(f"{a} không phải là số hạnh phúc.")
except ValueError:
    print("Lỗi gỡ lỗi nhập liệu: Bạn đã nhập sai định dạng (ValueError).")
```

```python
#W10a5
my_list = ["10", "abc", "20.5", "30", 40, [1, 2]]
result_list = []

for item in my_list:
    try:
        val = int(float(str(item)))
        result_list.append(val)
    except (ValueError, TypeError):
        print(f"Cảnh báo: Không thể chuyển đổi phần tử '{item}' sang int. Bỏ qua!")

print("List sau khi chuyển đổi:", result_list)
```

```python
#W10a6
import math
try:
    x = float(input("Nhập số x: "))
    if x < 0:
        raise ValueError("So am")
    print(f"{math.sqrt(x):.2f}")
except ValueError as e:
    if str(e) == "So am":
        print("So am")
    else:
        print("Lỗi: Dữ liệu nhập vào không phải là số.")
```

```python
#W10A10
try:
    N = int(input("Nhập số lượng phần tử N: "))
    if N <= 0:
        print("Mang khong hop le")
    else:
        arr = []
        for i in range(N):
            val = int(input(f"Nhập phần tử dương thứ {i+1}: "))
            if val <= 0:
                raise ValueError 
            arr.append(val)
        
        if len(arr) == len(set(arr)):
            print("Tổng các phần tử:", sum(arr))
        else:
            print("Mang khong hop le")
except ValueError:
    print("Mang khong hop le")
```
