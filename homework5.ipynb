```python
#W5A23
def quadratic_roots_count(a,b,c):
    delta=b**2-4*a*c
    if delta<0:
        return "vo nghiem"
    if delta>0:
        x=(-b+delta**(0.5))/(2*a)
        y=(-b-delta**(0.5))/(2*a)
        return x,y
    if delta ==0:
        z=-b/(2*a)
        return z
a,b,c=map(int,input().split())
print(quadratic_roots_count(a,b,c))
```

```python
#W5A25
x1,x2,x3,x4=map(int,input("nhap cac hoanh do:").split())
y1,y2,y3,y4=map(int,input("nhap cac tung do:").split())
p=list(map(int,input("nhap tao do p:").split()))
tapx=[x1,x2,x3,x4]
tapy=[y1,y2,y3,y4]
quad=list(zip(tapx,tapy))
def Stamgiac(a,b,c):
    S=0.5*(abs(a[0]*(b[1]-c[1])+b[0]*(c[1]-a[1])+c[0]*(a[1]-b[1])))
    return S
def check(quad,p):
    a=quad[0]
    b=quad[1]
    c=quad[2]
    d=quad[3]
    S1=Stamgiac(a,b,c)+Stamgiac(a,d,c)
    Sp1=Stamgiac(p,a,b)
    Sp2=Stamgiac(p,b,c)
    Sp3=Stamgiac(p,c,d)
    Sp4=Stamgiac(p,a,d)
    if (Sp1+Sp2+Sp3+Sp4)==S1:
        return "p co nam trong tu giac ABCD"
    else:
        return "p khong nam trong tu giac ABCD"
print(check(quad,p))

```

```python
#W5A28
import math
x1,x2,x3=map(int,input("nhap hoanh do x:").split())
y1,y2,y3=map(int,input("nhap tung do y:").split())
x=[x1,x2,x3]
y=[y1,y2,y3]
ABC=list(zip(x,y))
def tinh_goc(ABC):
    A=ABC[0]
    B=ABC[1]
    C=ABC[2]
    BAx=A[0]-B[0]
    BAy=A[1]-B[1]
    BCx=C[0]-B[0]
    BCy=C[1]-B[0]
    tichvohuong=BAx*BCx+BAy*BCy
    dodaiBA=math.sqrt(BAx**2+BAy**2)
    dodaiBC=math.sqrt(BCx**2+BCy**2)
    if dodaiBA==0 or dodaiBC==0:
        return 0 
    cos_ABC=tichvohuong/(dodaiBC*dodaiBA)
    gocrad=math.acos(cos_ABC)
    gocdo=math.degrees(gocrad)
    return round(gocdo,2)
print(f"goc ABC co toa do 3 diem A,B,C lan luot la{ABC} la:{tinh_goc(ABC)}")
```

```python
#W5A30
def nhithuc(n,k):
    if n>k:
        N=1
        for i in range(1,n+1):
           N=N*i
        K=1
        for j in range(1,k+1):
            K=K*j
        h=n-k
        H=1
        for l in range(1,h+1):
            H=H*l
        C=N/(K*H)
        return C
    else:
        return "error"
n,k=map(int,input().split())
print(nhithuc(n,k))


        
```

```python
#w5a33
def smart_taxi_fare(loaixe, thoigianbatdaudi, km, thoigiancho):
    tong = 0
    if loaixe == 4:
        gia = [10000, 13500, 11000, 800]
    elif loaixe == 7:
        gia = [15000, 17500, 14500, 1200]
    else:
        return "Loại xe không hợp lệ"

    if km <= 1:
        tong = gia[0]
    elif km <= 25:
        tong = gia[0] + (km - 1) * gia[1]
    else:
        tong = gia[0] + (24 * gia[1]) + (km - 25) * gia[2]

    tong += thoigiancho * gia[3]

    if not (5 < thoigianbatdaudi < 23):
        tong = tong * 1.1
    if tong>500000:
        tong=tong-20000

    return round(tong)

inp = input("Nhập (Loại_xe Giờ_đi KM Phút_chờ): ").split()
loaixe, thoigianbatdaudi, km, thoigiancho = map(float, inp)
    
ket_qua = smart_taxi_fare(int(loaixe), int(thoigianbatdaudi), km, thoigiancho)
print(f"Tổng tiền taxi là: {ket_qua:,} VNĐ")

```
