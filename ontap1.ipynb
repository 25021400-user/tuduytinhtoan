```python
#bai1
import math
def smart_taxi_fare(car_type,start_hour,km,wait_minutes):

    a=[22,23,0,1,2,3,4,5]
    if car_type==4:
        tongtien=11000
        if 0.8<km<=20:
            tongtien=tongtien+(km-0.8)*13000
        if km>20:
            tongtien=tongtien+(20-0.8)*13000+(km-20)*10500
    if car_type==7:
        tongtien=12000
        if 0.8<km<=20:
            tongtien=tongtien+(km-0.8)*14000
        if km>20:
            tongtien=tongtien+(20-0.8)*14000+(km-20)*12000
    if wait_minutes>5:
        tongtien=tongtien+(math.ceil(wait_minutes-5)/5)*2000
    if start_hour in a:
        tongtien=tongtien*1.2  
    if km>30 and start_hour not in a: 
        tongtien=tongtien*0.9
    if tongtien<50000 and km>5:
        tongtien=50000      
    return tongtien
car_type,start_hour,km,wait_minutes=map(float,input().split())
print(f"{smart_taxi_fare(car_type,start_hour,km,wait_minutes):.0f} VND")
```

```python
#16
import math
x1,x2,x3=map(int,input().split())
y1,y2,y3=map(int,input().split())

def goc(x1,x2,x3,y1,y2,y3):
    ba=[x2-x1,y2-y1]
    bc=[x3-x1,y3-y1]
    cos=(ba[0]*bc[0]+ba[1]*bc[1])/(((ba[0]**2+ba[1]**2)**0.5)*((ba[0]**2+ba[1]**2)**0.5))
    gocradian=math.acos(cos)
    gocdo=math.degrees(gocradian)
    return gocdo
print(f"{goc(x1,x2,x3,y1,y2,y3):.2f}")
```

```python
#19
def nhiphan(a):
    sonhiphan=""
    while a>0:
        c=a%2
        sonhiphan=str(c) + sonhiphan
        a=a//2
    return sonhiphan
def xor(a,b):
    x=nhiphan(a)
    y=nhiphan(b)
    if len(x)<=len(y):
        h=len(y)-len(x)
        for i in range(h):
            x="0"+x  
    if len(x)>=len(y):
        h=len(x)-len(y)
        for i in range(h):
            y="0"+y 
    s=""
    for i in range(len(x)):
        if x[i]==y[i]:
            s=s+"0"
        if x[i]!=y[i]:
            s=s+"1"
                
    tong=0
    for i in range(len(s)):
        tong=tong+int(s[i])*(2**(len(s)-1-i))
    return tong


a,b=map(int,input().split())
print(xor(a,b))
```

```python
#25
x1,x2,x3,x4=map(int,input("nhap cac hoanh do:").split())
y1,y2,y3,y4=map(int,input("nhap cac tung do:").split())
p=list(map(int,input("nhap tao do p:").split()))
tapx=[x1,x2,x3,x4]
tapy=[y1,y2,y3,y4]
quad=list(zip(tapx,tapy))
def Stamgiac(a,b,c):
    S=0.5*(abs(a[0]*(b[1]-c[1])+b[0]*(c[1]-a[1])+c[0]*(a[1]-b[1])))
    return S
def check(quad,p):
    a=quad[0]
    b=quad[1]
    c=quad[2]
    d=quad[3]
    S1=Stamgiac(a,b,c)+Stamgiac(a,d,c)
    Sp1=Stamgiac(p,a,b)
    Sp2=Stamgiac(p,b,c)
    Sp3=Stamgiac(p,c,d)
    Sp4=Stamgiac(p,a,d)
    if (Sp1+Sp2+Sp3+Sp4)==S1:
        return "p co nam trong tu giac ABCD"
    else:
        return "p khong nam trong tu giac ABCD"
print(check(quad,p))
```

```python
#26
def are_collinear(x1,y1,x2,y2,x3,y3):
    a=x2-x1
    b=y2-y1
    c=x3-x1
    d=y3-y1
    if (a*d-c*b)==0:
        return True
    else:
        return False
x1,y1,x2,y2,x3,y3=map(int,input().split())
print(are_collinear(x1,y1,x2,y2,x3,y3))
```
