```python
#w6a9
a=input().split()
b=input().split()
dic1={}
dic2={}
for i in a:
    key, value=i.split(":")
    dic1[key]=value
for i in b:
    key, value=i.split(":")
    dic2[key]=value
dic3={}
for i in dic1:
    if i in dic2:
        value=int(dic1[i])+int(dic2[i])
        dic3[i]=value
    if i not in dic2:
        dic3[i]=dic1[i]
for i in dic2:
    if i not in dic1:
        dic3[i]=dic2[i]
print(dic3)
```
