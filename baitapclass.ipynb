```python
#w8a2
class cylinder:
    def __init__(self, d, m, y):
        self.day = d
        self.month = m
        self.year = y

    def check_value(self):
        if self.year <= 0 or not (1 <= self.month <= 12):
            return False
            
        maxday = -1
        match self.month:
            case 1|3|5|7|8|10|12:
                maxday = 31
            case 4|6|9|11:
                maxday = 30
            case 2:
                if (self.year % 4 == 0 and self.year % 100 != 0) or (self.year % 400 == 0):
                    maxday = 29
                else:
                    maxday = 28
        return 1 <= self.day <= maxday

    def getnextday(self):
        maxday = -1
        match self.month:
            case 1|3|5|7|8|10|12: maxday = 31
            case 4|6|9|11: maxday = 30
            case 2:
                if (self.year % 4 == 0 and self.year % 100 != 0) or (self.year % 400 == 0):
                    maxday = 29
                else:
                    maxday = 28

       
        d = self.day + 1
        m = self.month
        y = self.year

        if d > maxday:
            d = 1
            m += 1
            if m > 12:
                m = 1
                y += 1

        
        return cylinder(d, m, y)

    def __str__(self):
        return f"{self.day:02d}/{self.month:02d}/{self.year}"

d, m, y = map(int, input().split('/'))
cylinder1 = cylinder(d, m, y)

if cylinder1.check_value():
    nextday = cylinder1.getnextday()
    print(nextday)
else:
    print("ngay nhap sai roi")
```

```python
#w8a3
class Point:
    def __init__(self,x,y) -> Point:
        self.x=x
        self.y=y
    def goc(self):
        return self.x==0 and self.y==0
    def ox(self):
        return self.y==0 and self.x!=0
    def oy(self):
        return self.x==0 and self.y!=0
          
    def khoangcach(self,other):
        return ((self.x-other.x)**2+(self.y-other.y)**2)**0.5
x,y=map(int,input().split())
p1 = Point(x,y)
p2 = Point(0,0)
if p1.goc():
    print("diem la goc toa do")
elif p1.ox():
    print("diem thuoc truc ox")
elif p1.oy():
    print("diem thuoc truc oy")
else:
    print("diem khong thoc ox,oy va goc tao do")
print(f"{p1.khoangcach(p2):.2f}")
   
```

```python
class Rectangle:
    def __init__(self, w, h):
        if float(w) <= 0 or float(h) <= 0:
            raise ValueError("Chiều rộng và chiều cao phải là số dương!")
        self.w = float(w)
        self.h = float(h)      
    def area(self):
        return self.w * self.h      
    def perimeter(self):
        return 2 * (self.w + self.h)      
    def scale(self, k):
        if k <= 0:
            raise ValueError("Hệ số tỷ lệ k phải lớn hơn 0!")
        self.w *= k
        self.h *= k

try:
    w, h, k = map(float, input("Nhập w, h, k cách nhau bằng khoảng trắng: ").split())
    rect = Rectangle(w, h)
    rect.scale(k)
    print(f"{rect.area():.2f}")
    print(f"{rect.perimeter():.2f}")
except ValueError as e:
    print(f"Lỗi dữ liệu: {e}")
```

```python
#w8a7
import math

class Fraction:
    def __init__(self, num, den):
        if den == 0:
          raise ZeroDivisionError("Mẫu số không thể bằng 0.") 
        g = math.gcd(num, den)
        self.num = num // g
        self.den = den // g
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den

    def __str__(self):
        if self.den == 1:
            return f"{self.num}"
        return f"{self.num}/{self.den}"
    def __add__(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)
        
    def __sub__(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)
        
    def __mul__(self, other):
        return Fraction(self.num * other.num, self.den * other.den)
        
    def __truediv__(self, other):
        if other.num == 0:
            raise ZeroDivisionError("Không thể chia cho phân số có tử số bằng 0.")
        return Fraction(self.num * other.den, self.den * other.num)


try:
    dong = input().split()
    f1 = Fraction(int(dong[0]), int(dong[1]))
    op = dong[2]
    f2 = Fraction(int(dong[3]), int(dong[4]))
    if op == '+': print(f1 + f2)
    elif op == '-': print(f1 - f2)
    elif op == '*': print(f1 * f2)
    elif op == '/': print(f1 / f2)
    else: print("Phép toán không hợp lệ")
except ZeroDivisionError as e:
    print(f"Lỗi toán học: {e}")
except (IndexError, ValueError):
    print("Định dạng nhập vào không đúng. Ví dụ: 1 2 + 3 4")
```

```python
#w8a8
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, x): self.balance += x 
    def withdraw(self, x):
        if x <= self.balance: self.balance -= x 

dong1 = input().split() 
owner = dong1[0] 
balance = float(dong1[1]) 

n = int(input()) 
tk = BankAccount(owner, balance)

for _ in range(n): 
    lenh = input().split() 
    if lenh[0] == "DEPOSIT": tk.deposit(float(lenh[1])) 
    elif lenh[0] == "WITHDRAW": tk.withdraw(float(lenh[1])) 

print(f"{tk.balance:.2f}") 
```

```python
#W8A9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = {}
    def add_score(self, subject, score):
        self.scores[subject] = score
    def avg(self):
        all_scores = self.scores.values()
        if not all_scores:
            return 0.0
        return sum(all_scores) / len(all_scores)
    def rank(self):
        dtb = self.avg()
        if dtb >= 8.0: return "Excellent"
        elif dtb >= 6.5: return "Good"
        elif dtb >= 5.0: return "Average"
        return "Poor"

ten = input().strip()
n = int(input())
sv = Student(ten)

for _ in range(n):
    dong = input().strip().rsplit(maxsplit=1)
    ten_mon = dong[0]
    diem_so = float(dong[1])
    sv.add_score(ten_mon, diem_so)

print(f"{sv.name} {sv.avg():.2f} {sv.rank()}")
```

```python
#w8a10
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = int(year)
class Library:
    def __init__(self):
        self.books = []
    def add(self, book): 
        self.books.append(book)  
    def count_by_author(self, name):
        return sum(1 for b in self.books if b.author == name)       
    def find_by_year(self, y):
        return sum(1 for b in self.books if b.year == int(y))
n = int(input())
lib = Library()
for _ in range(n):
    dong = input().strip()
    if " " in dong:
        lenh, noi_dung = dong.split(maxsplit=1)
    else:
        lenh, noi_dung = dong.replace(':', ' ').split(maxsplit=1)
    lenh = lenh.replace(':', '').strip()

    if lenh == "ADD":
        thong_tin = noi_dung.replace(':', ';').split(';')
        ten_sach = thong_tin[0].strip()
        tac_gia = thong_tin[1].strip()
        nam_xb = thong_tin[2].strip()   
        lib.add(Book(ten_sach, tac_gia, nam_xb))
        
    elif lenh == "COUNTYEAR":
        nam = noi_dung.replace(':', '').strip()
        print(lib.find_by_year(nam))
        
    elif lenh == "COUNT":
        tac_gia = noi_dung.replace(':', '').strip()
        print(lib.count_by_author(tac_gia))
```

```python
#w8a11
import math

class Vector3D:
    def __init__(self, x, y, z):
        self.x, self.y, self.z = x, y, z 
    def cross(self, o):
        return Vector3D(
            self.y * o.z - self.z * o.y, 
            self.z * o.x - self.x * o.z, 
            self.x * o.y - self.y * o.x
        )        
    def dot(self, o): 
        return self.x * o.x + self.y * o.y + self.z * o.z        
    def length(self): 
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)
def read_pt():
    v = list(map(float, input().split()))
    return Vector3D(v[0], v[1], v[2])
A, B, C, D = read_pt(), read_pt(), read_pt(), read_pt()
AB = Vector3D(B.x - A.x, B.y - A.y, B.z - A.z)
BC = Vector3D(C.x - B.x, C.y - B.y, C.z - B.z)
CD = Vector3D(D.x - C.x, D.y - C.y, D.z - C.z)
N1 = AB.cross(BC)
N2 = BC.cross(CD)
len_N1 = N1.length()
len_N2 = N2.length()
if len_N1 == 0 or len_N2 == 0:
    print("Khong the xac dinh goc (cac diem thang hang)")
else:
    cos_phi = abs(N1.dot(N2)) / (len_N1 * len_N2)
    cos_phi = max(-1.0, min(1.0, cos_phi))
    phi = math.degrees(math.acos(cos_phi))
    print(f"{phi:.2f}")
```

```python
#w8a12

class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        self.real = float(realPart)
        self.imag = float(imaginaryPart)

  
    def __mul__(self, other):
       
        r = self.real * other.real - self.imag * other.imag
        i = self.real * other.imag + self.imag * other.real
        return ComplexNumber(r, i)

    def __str__(self):
        r_int = int(round(self.real))
        i_int = int(round(self.imag))
        return f"{r_int} {i_int}"


d1 = input().split()
d2 = input().split()

sp1 = ComplexNumber(d1[0], d1[1])
sp2 = ComplexNumber(d2[0], d2[1])


kq = sp1 * sp2

print(kq)
```

```python
#W8A13
import math

class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.e1 = float(edge1)
        self.e2 = float(edge2)
        self.e3 = float(edge3)

    def is_valid(self):
    
        a, b, c = self.e1, self.e2, self.e3
        return a + b > c and a + c > b and b + c > a

    def get_area(self):   
        a, b, c = self.e1, self.e2, self.e3
        p = (a + b + c) / 2
        return math.sqrt(p * (p - a) * (p - b) * (p - c))


dong = list(map(int, input().split()))
tg = Triangle(dong[0], dong[1], dong[2])

if tg.is_valid():
    dt = tg.get_area()
    print(f"{dt:.2f}")
else:
    print("invalid")
```

```python
#w8a15
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = float(salary)        
    def income(self): 
        return self.salary
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = float(bonus)
        
    def income(self): 
        return self.salary + self.bonus
class Director(Employee):
    def __init__(self, name, salary, bonus, allowance):
        super().__init__(name, salary)
        self.bonus = float(bonus)
        self.allowance = float(allowance)
        
    def income(self):
        return self.salary + self.bonus + self.allowance

loai = input().strip()
dong2 = input().split()
cls = globals()[loai]
nhan_vien = cls(*dong2)
print(f"{nhan_vien.income():.0f}")
```
