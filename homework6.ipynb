```python
#w6a17
n=int(input("nhap so hang va so cot:"))
matran=[]
for i in range(n):
    hang=[]
    for j in range(n):
        so=int(input())
        hang.append(so)
    matran.append(hang)
dcc=[]
dcp=[]
for i in range(n):
    for j in range(n):
        if i==j:
            c=matran[i][j]
            dcc.append(c)
for i in range(n):
    d=n-1-i
    c=matran[i]
    e=c[d]
    dcp.append(e)

print("\nKết quả ma trận:")
for hang in matran:
    for so in hang:
        print(f"{so:>4}", end="")
    print()
print(f"đường chéo chính:{dcc}")
print(f"đường chéo phụ:{dcp}")
```

```python
#w6a18
n,m,k=map(int,input("nhap so hang,so cot, so nguyen k:").split())
print(f"k={k}")
matran=[]
for i in range(n) :
    hang=[]
    for j in range(m) :
        so=int(input())
        hang.append(so)
    matran.append(hang)
print("\nKết quả ma trận:")
for hang in matran:
    for so in hang:
        print(f"{so:>4}", end="")
    print()
if k>n and k>m:
    print(f"khong co cot {k} va hang {k}")
elif k<n and k>m:
    c=matran[k]
    tong=0
    for i in c:
        tong+=i
    print(f"tong hang thu k:{tong}")
elif k>=n and k<=m:
    tong=0
    for i in range(n):
        c=matran[i]
        a=c[k]
        tong+=a
    print(f"tong cot thu k:{tong}")
elif  k<n and k<m:
    c=matran[k]
    tonghang=0
    for i in c:
        tonghang+=i
    tongcot=0
    for i in range(n):
        b=matran[i]
        a=b[k]
        tongcot+=a
    print(f"tong cot k va hang k lan luot la :{tongcot}, {tonghang} ")
```

```python
#w6a19
def maxlist(a):
    b=a[0]
    c=[a[0]]
    for i in a:
        if i>=b:
           b=i
           c.pop(0)
           c.append(i)
    d=c[0]
    return d
a=list(map(int,input().split()))
print(a)
print(maxlist(a))
```

```python
#w6a28
a=list(map(int,input().split()))
dic={}
for i in range(1,len(a)-1):
    b=a[i]-a[i-1]
    c=a[i+1]-a[i]
    dic[b]=c
x,y,z=0,0,0
for key,value in dic.items():
    if int(key)>0 and int(value)>0:
        x=1
    elif int(key)<0 and int(value)<0:
        y=1
    else:
        z=1    
if x==1 and y==0 and z==0:
    print(a)
    print("chuoi tang")
if x==0 and y==1 and z==0:
    print(a)
    print("chuoi giam")
if z==1:
    print(a)
    print("chuoi vo danh ")
```
