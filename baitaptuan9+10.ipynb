```python
#w9a7
flowerbed=list(map(int,input("nhâp chỉ 0 với 1 ").split()))
k=int(input("soluonghoa:"))
def canPlaceFlowers(flowerbed, k):
    c=len(flowerbed)
    for i in range(c):
        if flowerbed[i] == 0:
            left_empty = (i == 0) or (flowerbed[i - 1] == 0)
            right_empty = (i == c - 1) or (flowerbed[i + 1] == 0)
            if left_empty and right_empty:
                flowerbed[i] = 1 
                k -= 1           
    return k<=0
print(canPlaceFlowers(flowerbed, k))



```

```python
#w9a8
moves=input("nhập bước di chuyển(R,L,U,D):")
da_qua=[[0,0]]
def is_path_crossing(moves):
    x=0
    y=0
    for i in moves:
        if i=="R":
            x+=1
        if i=="L":
            x-=1
        if i=="U":
            y+=1
        if i=="D":
            y-=1
        c=[x,y]
        if c in da_qua:
            return True
        da_qua.append(c)
    return False

    
print(is_path_crossing(moves))

```

```python
#w9a9
nums = list(map(int, input("Nhập 5 số nguyên: ").split()))

def Median(nums):
    def sort_pair(i, j):
        if nums[i] > nums[j]:
            nums[i], nums[j] = nums[j], nums[i]       
    sort_pair(0, 1)
    sort_pair(2, 3)  
    if nums[0] > nums[2]:
        nums[0], nums[2] = nums[2], nums[0]
        nums[1], nums[3] = nums[3], nums[1]
    nums[0] = nums[4]
    sort_pair(0, 1)   
    if nums[0] > nums[2]:
        nums[0], nums[2] = nums[2], nums[0]
        nums[1], nums[3] = nums[3], nums[1]
    if nums[1] < nums[2]:
        return nums[1]
    else:
        return nums[2]

print("Mảng ban đầu:", nums)
print("Số trung vị là:", Median(nums))

    

```

```python
#w9a10
def findRadius(houses, heaters):
    houses.sort()
    heaters.sort() 
    max_radius = 0
    j = 0     
    for i in range(len(houses)):
        while j < len(heaters) - 1 and abs(heaters[j + 1] - houses[i]) <= abs(heaters[j] - houses[i]):
            j += 1
        current_dist = abs(heaters[j] - houses[i])
        max_radius = max(max_radius, current_dist)
    return max_radius
houses = list(map(int, input("Nhập vị trí các ngôi nhà (cách nhau bởi khoảng trắng): ").split()))
heaters = list(map(int, input("Nhập vị trí các lò sưởi (cách nhau bởi khoảng trắng): ").split()))
if not houses or not heaters:
        print("Dữ liệu đầu vào không hợp lệ (Mảng không được trống)!")
else:      
    result = findRadius(houses, heaters)
    print(f"\n=> Bán kính nhỏ nhất cần thiết của các lò sưởi là: {result}")
```

```python
#w9a13
def power(base, exp):
    if exp==0:
        return 1
    if exp==1:
        return base
    if exp < 0:
        return 1 / power(base, -exp)
    a = power(base, exp // 2)
    if exp % 2 == 0:
        return a * a
    else:
        return base * a * a
print(power(2, 10))   
print(power(5, 3))    
print(power(3, -2))
    
```

```python
#w9a15
def kiem_tra_tang_nghiem_ngat(a):
    if len(a) < 2:
        return True
    for i in range(1, len(a)):
        if a[i] <= a[i-1]:
            return False
    return True
def kiem_tra_giam_nghiem_ngat(a):
    if len(a) < 2:
        return True      
    for i in range(1, len(a)):
        if a[i] >= a[i-1]:
            return False
    return True

a = list(map(int, input("Nhập dãy số nguyên (cách nhau bởi dấu cách): ").split()))
print("Mảng đã nhập:", a)

is_tang = kiem_tra_tang_nghiem_ngat(a)
is_giam = kiem_tra_giam_nghiem_ngat(a)
if is_tang:
    print("-> Đây là dãy số TĂNG nghiêm ngặt.")
elif is_giam:
    print("-> Đây là dãy số GIẢM nghiêm ngặt.")
else:
    print("-> Dãy số này KHÔNG tăng cũng KHÔNG giảm nghiêm ngặt.")
```
