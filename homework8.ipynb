```python
#W8A16
class Shape:
    def area(self):
        return 0.0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h        
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r        
    def area(self):
        return 3.14 * (self.r ** 2)
shape_type = input().strip()
params = input().strip().split()
if shape_type == "RECTANGLE":
    w = float(params[0])
    h = float(params[1])
    obj = Rectangle(w, h)
    print(f"{obj.area():.2f}")
elif shape_type == "CIRCLE":
    r = float(params[0])
    obj = Circle(r)
    print(f"{obj.area():.2f}")
```

```python
#W8A17
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed        
    def info(self):
        return f"{self.brand}-{self.speed}"
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
        
    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"


obj_type = input().strip()
data = input().strip().split()

if obj_type == "Vehicle":
    brand = data[0]
    speed = data[1]
    obj = Vehicle(brand, speed)
    print(obj.info())
elif obj_type == "Car":
    brand = data[0]
    speed = data[1]
    seats = data[2]
    obj = Car(brand, speed, seats)
    print(obj.info())
```

```python
#W8A18
class Polygon:
    def perimeter(self):
        return 0

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def perimeter(self):
        return self.a + self.b + self.c


a, b, c = map(int, input().strip().split())
triangle = Triangle(a, b, c)
print(triangle.perimeter())
```

```python
#W8A19
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def describe(self):
        return f"{self.name} {self.age}"

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school
        
    def describe(self):
        return f"{self.name} {self.age} {self.school}"


obj_type = input().strip()
data = input().strip().split()

if obj_type == "Person":
    name = data[0]
    age = data[1]
    obj = Person(name, age)
    print(obj.describe())
elif obj_type == "Student":
    name = data[0]
    age = data[1]
    school = data[2]
    obj = Student(name, age, school)
    print(obj.describe())
```

```python
#W8A20
class Account:
    def __init__(self, balance):
        self.balance = balance
        
    def deposit(self, x):
        self.balance += x
        
    def get_balance(self):
        return self.balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
        
    def add_interest(self):
        self.balance += self.balance * self.rate


balance, rate = map(float, input().strip().split())
account = SavingAccount(balance, rate)
account.add_interest()
print(f"{account.get_balance():.2f}")
```

```python
#W8A21
class Printer:
    def print_value(self, x):
        print(x)

class BinaryPrinter(Printer):
    def print_value(self, x):
        print(bin(x)[2:])

n = int(input().strip())
printer = BinaryPrinter()
printer.print_value(n)
```

```python
#w8a22
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def trace(self):
        return self.a + self.d
        
    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)


a, b, c, d = map(int, input().strip().split())
matrix = Matrix2x2(a, b, c, d)
print(f"{matrix.trace()} {matrix.determinant()}")
```

```python
#w8a23
class Polynomial:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def evaluate(self, x):
        return self.a * x + self.b


a, b, x = map(float, input().strip().split())
poly = Polynomial(a, b)
result = poly.evaluate(x)
print(int(result) if result.is_integer() else result)
```

```python
#w8a24
class Polynomial2:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def evaluate(self, x):
        return self.a * (x ** 2) + self.b * x + self.c


a, b, c, x = map(float, input().strip().split())
poly2 = Polynomial2(a, b, c)
result = poly2.evaluate(x)
print(int(result) if result.is_integer() else result)
```

```python
#w8a25
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        return f"{k}x + {m}"

a, b, c, d = map(int, input().strip().split())
poly_sum = PolynomialSum(a, b, c, d)
print(poly_sum.sum_str())
```

```python
#w8a27
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent
        
    def final_price(self):
        return self.price * (100 - self.percent) / 100


data = input().strip().split()
name = data[0]
price = float(data[1])
percent = float(data[2])

product = DiscountProduct(name, price, percent)
print(f"{product.final_price():.2f}")
```

```python
#w8a28
class Counter:
    def __init__(self):
        self.count = 0
        
    def inc(self):
        self.count += 1
        
    def dec(self):
        self.count -= 1
        
    def value(self):
        return self.count


n = int(input().strip())
operations = input().strip().split()

counter = Counter()
for op in operations:
    if op == "INC":
        counter.inc()
    elif op == "DEC":
        counter.dec()

print(counter.value())
```

```python
#w8a29
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
        
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s


h, m, s = map(int, input().strip().split())
t = Time(h, m, s)
print(t.to_seconds())
```

```python
#W8a30
class Stack:
    def __init__(self):
        self.elements = []
        
    def push(self, x):
        self.elements.append(x)
        
    def pop(self):
        if len(self.elements) > 0:
            return self.elements.pop()
        
    def top(self):
        if len(self.elements) > 0:
            return self.elements[-1]


n = int(input().strip())
stack = Stack()

for _ in range(n):
    command = input().strip().split()
    if command[0] == "PUSH":
        stack.push(command[1])
    elif command[0] == "POP":
        stack.pop()
    elif command[0] == "TOP":
        print(stack.top())
```
