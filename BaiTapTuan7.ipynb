```python
#w7a1
a=list(map(int,input().split()))
socantim=int(input())
vitridau=0
vitricuoi=len(a)-1
def  binary_search(a,vitridau,vitricuoi,socantim):
    while vitridau<=vitricuoi:
        vitrigiua=vitridau+(vitricuoi-vitridau)//2
        if a[vitrigiua]==socantim:
            return vitrigiua
        elif a[vitrigiua]<socantim:
             vitridau= vitrigiua+1
        else:
            vitricuoi=vitrigiua-1
    return -1
print(a)
print(socantim)
print(binary_search(a,vitridau,vitricuoi,socantim)) 
```

```python
#w7a2
a=list(map(int,input().split()))
x=int(input())
def count_occurrences(a, x):
    count = 0
    for num in a:
        if num == x:
            count += 1
    return count
print(a)
print(x)
print(count_occurrences(a, x))
```

```python

#w7a4
a=list(map(int,input().split()))
def slanmax(a):
    if not a:
        return None, 0
    somax = a[0]
    solanmax = 0
    for i in a:
        solan = 0
        for j in a:
            if i == j:
                solan += 1
        if solan > solanmax:
            solanmax = solan
            somax = i
    return somax, solanmax

print(a)
print(slanmax(a))
              
```

```python
#w7a7
a=list(map(int,input().split()))
def length_of_lis(nums):
    danh_sach_tam = []   
    for x in nums:
        if not danh_sach_tam or x > danh_sach_tam[-1]:
            danh_sach_tam.append(x)
        else:
            for i in range(len(danh_sach_tam)):
                if danh_sach_tam[i] >= x:
                    danh_sach_tam[i] = x
                    break
                    
    return len(danh_sach_tam)
print(a)
print(length_of_lis(a))
```

```python
#w7a8
ds1=[]
print("/n=== tạo danh sách ===")
print("1.tạo danh sách co dạng [l,r] với l<=r")
print("2.thoát")
i=-1
while i<0:
    choice=int(input("nhập lựa chọn: "))
    match choice:
        case 1:
            l, r = map(int, input("nhập khoảng [l, r]: ").split())
            ds1.append([l, r])
        case 2:
            break
q=list(map(int,input("nhập số phần tử cần kiểm tra: ").split()))
ds2=[]
min=ds1[0][1]-ds1[0][0]+1
for i in ds1:
    a= i[1] - i[0]+1
    if i in ds2:
        continue
    if a <= min:
        min = a
        ds2.insert(0, i)
    if a > min:
        b=False
        for j in range(len(ds2)):
            if ds2[j][1] - ds2[j][0]+1 > a:
                ds2.insert(j, i)
                b=True
                break
            if not b and i not in ds2:
                ds2.append(i)
ds3=[]
for i in q:
    for j in ds2:
        if j[0] <= i <= j[1]:
            a= j[1] - j[0]+1
            ds3.append(a)
            break
print("danh sách ban đầu:", ds1)    
print(q)
print("danh sách độ dài nhỏ nhất các khoảng chứa các phần tử:", q, ds3)
```

```python
#7a12
A = input("Nhập A: ")
B = input("Nhập B: ")
C = input("Nhập C: ")
D = input("Nhập D: ")

a = [A, B, C, D]
print(a)
ngay_trong_tuan = ["Chu nhat", "Thu hai", "Thu ba", "Thu tu", "Thu nam", "Thu sau", "Thu bay"]
dic = {A: "A", B: "B", C: "C", D: "D"}

la_ngay = any(x in ngay_trong_tuan for x in a)

c = []

if la_ngay:
    for ngay in ngay_trong_tuan:
        if ngay in a:
            c.append(dic[ngay])
else:
    a.sort()
    for i in a:
        c.append(dic[i])

for i in c:
    print(i, end=" ")
        
```
