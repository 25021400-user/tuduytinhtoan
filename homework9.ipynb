```python
#w9a16
def generate_binary(n, m, current_str=""):
    if n == 0 and m == 0:
        print(current_str)
        return
    
    if n > 0:
        generate_binary(n - 1, m, current_str + "0")
    

    if m > 0:
        generate_binary(n, m - 1, current_str + "1")


n, m = map(int,input().split())
generate_binary(n, m)
```

```python
#w9a17
def lis(arr):
    n = len(arr)
    if n == 0: return 0
    L = [1] * n    
    for i in range(1, n):
        for j in range(0, i):
            if arr[i] > arr[j]:
                L[i] = max(L[i], L[j] + 1)           
    return max(L)


arr = list(map(int, input("Nhập dãy số (cách nhau bởi khoảng trắng): ").split()))
print("Độ dài dãy con tăng nghiêm ngặt dài nhất:", lis(arr))
```

```python
#w9a18
def count_pairs(arr, k):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == k:
                count += 1
    return count

arr = list(map(int, input("Nhập dãy số: ").split()))
k = int(input("Nhập số k: "))
print("Số cặp có tổng bằng k là:", count_pairs(arr, k))
```

```python
#w9a19
def remove_duplicates(arr):
    result = []
    for x in arr:
        if x not in result:
            result.append(x)
    return result

arr = list(input().split())
print("Mảng sau khi lọc trùng:", remove_duplicates(arr))
```

```python
#w9a20
def clean_and_capitalize(s):
    words = []
    current_word = ""
    for char in s:
        if char != " ":
            current_word += char
        else:
            if current_word != "":
                words.append(current_word)
                current_word = ""
    if current_word != "":
        words.append(current_word)
    
    res_string = ""
    for i in range(len(words)):
        word = words[i]
        first_char = word[0]
        if 'a' <= first_char <= 'z':
            first_char = chr(ord(first_char) - 32)
            
        cleaned_word = first_char + word[1:]
        
        
        if i == 0:
            res_string += cleaned_word
        else:
            res_string += " " + cleaned_word
            
    return res_string

s = "   nguYen   vAn    a   "
print(f"Kết quả: '{clean_and_capitalize(s)}'")
```

```python
#w9a21
def quickselect(arr, k):
    target = len(arr) - k
    def select(left, right):
        pivot = arr[right]
        i = left
        for j in range(left, right):
            if arr[j] <= pivot:
                arr[i], arr[j] = arr[j], arr[i]
                i += 1
        arr[i], arr[right] = arr[right], arr[i]
        
        if i == target:
            return arr[i]
        elif i < target:
            return select(i + 1, right)
        else:
            return select(left, i - 1)

    return select(0, len(arr) - 1)

arr = [3, 2, 1, 5, 6, 4]
k = 2 
print(f"Phần tử lớn thứ {k} là:", quickselect(arr, k))
```

```python
#w9a22
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False
        
    count1 = {}
    count2 = {}
    
    for char in s1:
        count1[char] = count1.get(char, 0) + 1
    for char in s2:
        count2[char] = count2.get(char, 0) + 1
        
    return count1 == count2

print(is_anagram("listen", "silent")) 
```

```python
#w9a23
def largest_number(s, k):
    stack = []
    to_remove = len(s) - k 
    for char in s:
        while to_remove > 0 and stack and stack[-1] < char:
            stack.pop()
            to_remove -= 1
        stack.append(char)
        
    return "".join(stack[:k])

s = "78934340999"
k = 7
print("Số lớn nhất thu được:", largest_number(s, k)) 
```

```python
#w9a24
def compress_string(s):
    if not s: return ""
    
    result = ""
    count = 1
    
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result += s[i-1] + str(count)
            count = 1
            
    result += s[-1] + str(count)
    return result

s = "aaaaaafdhhhhh"
print("Chuỗi sau khi nén:", compress_string(s)) 
```

```python
#w9a25
def is_strong_password(pwd):
    if len(pwd) < 8:
        return False
        
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    special_chars = "!@#$"
    
    for char in pwd:
        if 'a' <= char <= 'z':
            has_lower = True
        elif 'A' <= char <= 'Z':
            has_upper = True
        elif '0' <= char <= '9':
            has_digit = True
        elif char in special_chars:
            has_special = True
            
    return has_lower and has_upper and has_digit and has_special

pwd = input("Nhập mật khẩu kiểm tra: ")
if is_strong_password(pwd):
    print("Mật khẩu MẠNH")
else:
    print("Mật khẩu YẾU")
```
